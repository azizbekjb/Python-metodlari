<h2 id="file-close">1. close()</h2>

---
**Tushuntirish**: Ochilgan faylni yopadi. Har doim ochilgan faylni vazifa tugagandan so'ng yopish talab etiladi.


**Sintaksis**:
```
fayl.close()
```

````python
# Oddiy hol
fayl = open("./fayllar/1-fayl.txt", "r")
print(fayl.read())
fayl.close()
````

<h2 id="file-fileno">2. fileno()</h2>

---

**Tushuntirish**: fileno() metodi oqimning fayl deskriptorini son ko‘rinishida qaytaradi.

Agar operatsion tizim fayl deskriptoridan foydalanmasa, xatolik yuzaga keladi.

**Sintaksis**:
```
fayl.fileno()
```

````python
fayl = open("./fayllar/1-fayl.txt", "r")
print(fayl.read())
print(fayl.fileno())
fayl.close()
````

<h2 id="file-flush">3. flush()</h2>

---

**Tushuntirish**: Bu metodi ichki buferni bo‘shatadi. Ya'ni faylga biror ma;lumot yozayoganimizda u avva buffer(vaqtinchalik xotirada)da saqlanishi mumkin. Bu metod esa bufferda yig'ilgan ma'lumotlarni faylga yozishni ta'minlaydi.

**Sintaksis**:
```
fayl.flush()
```

````python
fayl = open("./fayllar/1-fayl.txt", "a")
fayl.write("Faylga yangi qator qo'shildi.")
fayl.flush()
fayl.write("... yana bir boshqa qator.")
fayl.close()
````

````python
fayl = open("./fayllar/1-fayl.txt", "r")
print(fayl.read())
````
