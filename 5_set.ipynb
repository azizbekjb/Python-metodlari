<h2 id="set-add">1. add()</h2>

---
**Tushuntirish**: To'plamga yangi element qo'shadi. Agar element to'plamda mavjud bo'lsa, element qo'shilmaydi.

**Sintaksis**:
```
toplam.add(element)
```

Bu yerda:
- **element**: Majburiy parametr. Qo'shish uchun.

````python
# Oddiy hol: Mavjud bo'lmagan elementni 
# qo'shish
toplam = {1, 'a', 'b'}
element = 'c'
toplam.add(element)
print(toplam)
````

````python
# Maxsus hol: Mavjud bo'lgan elementni 
# qo'shish
toplam = {1, 'a', 'b'}
element = 'a'
toplam.add(element)
print(toplam)

# Tushuntirish:
#   - 'a' elementi to'plamda mavjud.
#   - Shu sababli element qo'shilmadi.
````

<h2 id="set-clear">2. clear()</h2>

---
**Tushuntirish**: To'plamdagi barcha elementlarni o'chiradi. Hech qanday qiymat qaytarmaydi.

**Sintaksis**:
```
toplam.clear()
```


````python
# Oddiy hol
toplam = {1, 2, 4, 5}
toplam.clear()
print(toplam)
````

<h2 id="set-copy">3. copy()</h2>

---

**Tushuntirish**: To'plamning nusxasini qaytaradi.

**Sintaksis**:
```
toplam.copy()
```


````python
toplam = {1, 2, 3, 4}
x = toplam.copy()
print(x)
````

<h2 id="set-difference">4. difference()</h2>

---

**Tushuntirish**: Berilgan to'plamlar o'rtasidagi ayirmani qaytaradi. Qaytarilgan to'plam elementlari birinchi to'plamda mavjud va ikkalsida mavjud bo'lmagan elementlardan tashkil topadi. Qisqartirib ishlatish uchun ```-``` belgisidan foydalanish mumkin.

**Umumiy sintaksis**:
```
toplam.difference(toplam2, toplam3, ... )
```
