<h2 id="set-add">1. add()</h2>

**Sintaksis**:
```
toplam.add(element)
```

Bu yerda:
- **element**: Majburiy parametr. Qo'shish uchun.

````python
# Oddiy hol: Mavjud bo'lmagan elementni 
# qo'shish
toplam = {1, 'a', 'b'}
element = 'c'
toplam.add(element)
print(toplam)
````

<h2 id="set-difference">4. difference()</h2>

---

**Tushuntirish**: Berilgan to'plamlar o'rtasidagi ayirmani qaytaradi. Qaytarilgan to'plam elementlari birinchi to'plamda mavjud va ikkalsida mavjud bo'lmagan elementlardan tashkil topadi. Qisqartirib ishlatish uchun ```-``` belgisidan foydalanish mumkin.

**Umumiy sintaksis**:
```
toplam.difference(toplam2, toplam3, ... )
```

Bu yerda:
- **toplam**: Majburiy parametr.
- **toplam2**, **toplam3**, ...: Ixtiyoriy parametrlar. Asosiy to'plamning boshqa to'plamlar o'rtasidagi ayirmani hisoblashi uchun.

**Maxsus sintaksis**:
```
toplam1 - toplam2 - toplam3 - ...
```



````python
# Oddiy hol: x to'plamda mavjud, lekin
# y to'plamda mavjud bo'lmagan eleemtlar

x = {"Salom", "qadrli", "Python", "dasturchilari"}
y = {"Mehri", "daryo", "va", "qadrli", "dasturchilar"}
z = x.difference(y) # yoki z = x - y
print(z)
````

````python
x - y
````

````python
# Maxsus hol: 2 va undan ortiq to'plamlarni
# birlashtirish


a = {"apple", "banana", "cherry"}
b = {"google", "microsoft", "apple"}
c = {"cherry", "micra", "bluebird"}

myset = a.difference(b, c)

print(myset)

# Tushuntirish:
#   - Dastlab a va b to'plamlar o'rtasigi ayirmani
#     hisoblaydi. Natija: {"banana", "cherry"}.
#   - So'ngra chiqgan natija bilan c to'plam 
#     o'rtasidagi ayirma hisoblanadi. Natija: 
#     {"banana"}
````

<h2 id="set-difference_update">5. difference_update()</h2>

---

**Tushuntirish**: Ikkila to'plam mavjud bo'lgan elementlarni olib tashlaydi. <a href="#set-difference">difference</a> metodidan farq qiladi, chunki <a href="#set-difference">difference</a> ikkala to'plamda mavjud bo'lgan elementlarni o'chirib, yangi to'plam qaytarardi. Bu metod esa ikkala to'plamda mavjud bo'lgan elementlarni haqiqy to'plamdan o'chiradi va yangidan saqlaydi. Metod nomi o'rniga qisqcha qilib ```-=``` belgilarini ishlatish mumkin.

**Umumiy sintaksis**:
```
toplam1.difference_update(toplam2, toplam3, ... )
```

**Maxsus sintaksis**:
```
toplam1 -= toplam2 | toplam3 | ...
```

Bu yerda:
- **toplam1**: Majburiy parametr.
- **toplam2**, **toplam3**, ...: Ixtiyoriy parametrlar. Asosiy to'plamning boshqa to'plamlar o'rtasidagi ayirmani hisoblashi uchun.

````python
# Oddiy hol: Xuddi avvalgi metoddagi 1-misol kabi. Faqat natija
# haqiqiy to'plamda o'zgaradi va hech qanday qiymat qaytarmaydi.
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"} 

x.difference_update(y) # yoki x -= y

print(x)
````

````python
# Maxsus hol: 2 va undan ortiq to'plamlarni birlashtirish

a = {"apple", "banana", "cherry"}
b = {"google", "microsoft", "apple"}
c = {"cherry", "micra", "bluebird"}

a.difference_update(b, c) # yoki a -= b | c

print(a)

# Tushuntirish:
#   - Dastlab a va b to'plamlar o'rtasigi ayirmani
#     hisoblaydi. Natija: {"banana", "cherry"}.
#   - So'ngra chiqgan natija bilan c to'plam 
#     o'rtasidagi ayirma hisoblanadi. Natija: 
#     {"banana"}
````

<h2 id="set-intersection">7. intersection</h2>

---
**Tushuntirish**: Ikki va undan ortiq to'plamlar o'rtasidagi kesishmani hisoblaydi. Metod nomi o'rniga qisqa qilib ```&``` belgisni ishlatish mumkin.

**Umumiy sintaksis**:
```
toplam.intersection(toplam2, toplam3, ... )
```

Bu yerda:
- **toplam**: Majburiy parametr.
- **toplam2**, **toplam3**, ...: Ixtiyoriy parametrlar. Asosiy to'plamning boshqa to'plamlar o'rtasidagi kesishmani hisoblashi uchun.

**Maxsus sintaksis**:
```
toplam1 & toplam2 & toplam3 & ...

````python
# Oddiy hol: Ikkit to'plam o'rtasidagi kisishmani hisoblash.
mevalar1 = {"olma", "anor", "nok"}
mevalar2 = {"olma", "olcha", "shaftoli"}
mavjud = mevalar1.intersection(mevalar2) # yoki mevalar1 & mevalar2
print(mavjud)
````

````python
# Maxsus hol: 2 va unda ortiq to'plar o'rtasidagi kesishmani
# hisoblash:

x = {"a", "b", "c"}
y = {"c", "d", "e"}
z = {"f", "g", "c"}

natija = x.intersection(y, z)

print(natija)

# Tushuntirish:
#   - Dastlab x va y to'plamlar o'rtasigi keshishmani
#     hisoblaydi. Natija: {"c"}
#   - So'ngra chiqgan natija bilan z to'plam 
#     o'rtasidagi kesishma hisoblanadi. Natija: 
#     {"c"}
````

<h2 id="set-intersection_update">8. intersection_update()</h2>

---

**Tushuntirish**: Ikki va undan ortiq to'plamlar o'rtasidagi kesishmani hisoblaydi. <a href="#set-intersection">intersection</a> metodidan ishlash tartibi bo'yicha farq qiladi. Sababi  <a href="#set-intersection">intersection</a> metodi yangi to'plam qaytaradi, bu metod esa yangi natijani birinchi to'plam o'rniga joylaydi. Metod nomi o'rniga maxsus ```&=``` belgisidan foydalanish mumkin.

**Umumiy sintaksis**:
```
tomplam1.intersection_update(toplam2, toplam3, ...)
```

**Maxsus sintaksis**:
```
toplam1 &= toplam2 & toplam3 & toplam4 & ...
```

Bu yerda:
- **toplam1**: Majburiy parametr.
- **toplam2**, **toplam3**, ...: Ixtiyoriy parametrlar. Asosiy to'plamning boshqa to'plamlar o'rtasidagi kesishmani hisoblashi uchun.

````python
# Oddiy hol: Ikki to'plam o'rtasidagi kesishma
x = {1, 2, 6, 3}
y = {5, 6, 1, 8}

print("Eski x: {0}".format(x))
x.intersection_update(y)
print("Yangi x: {0}".format(x))
````

````python
# Maxsus usul: Ikki to'plam o'rtasidagi kesishma.
# Belgilar orqali
x = {1, 2, 6, 3}
y = {5, 6, 1, 8}

print("Eski x: {0}".format(x))
x &= y
print("Yangi x: {0}".format(x))
````

````python
# Oddiy hol: Uchta to'plam o'rtasigi kesishma.
x = {1, 2, 6, 3}
y = {5, 6, 1, 8}
z = {1, 2, 3, 4}

print("Eski x: {0}".format(x))
x.intersection_update(y, z)
print("Yangi x: {0}".format(x))

# -------------------------------------------------------------------------
# Tushuntirish:
#   - Dastlab x va y to'plamlar o'rtasidagi kesishma hisoblanmoqda. 
#   - Keyin hosil bo'lgan yangi to'plam va z to'plam o'rtasidagi kesishma 
#     hisoblanmoqda.
#   - Yakuniy natija qaytadan x o'rniga qo'yilmoqda.
````

<h2 id="set-isdisjoint">9. isdisjoint()</h2>

---
**Tushuntirish**: Agar ikkala to'plamning elementlari turlicha bo'lsa ```True```(Rost), aks holda ```False```(yolg'on) qiymat qaytaradi.

**Sintaksis**:
```
toplam.isdisjoint(toplam2)
```

Bu yerda:
- **toplam**: Majburiy parametr. 
- **tomplam2**: Majburiy parametr.

````python
# Oddiy hol:
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "facebook"}

z = x.isdisjoint(y)

print(z) 

# -------------------------------------------------------------------------
# Tushuntirish:
#   - True qaytarmoqda, sababi to'plam elementlari turlicha.
````

````python
# Maxsus hol:
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.isdisjoint(y)

print(z) 

# -------------------------------------------------------------------------
# Tushuntirish:
#   - False qaytarmoqda, sababi to'plamda bir xil element bor. Ya'ni 
#     "apple" elementi ikkalasida ham mavjud.
````

<h2 id="set-issubset">10. issubset()</h2>

---
**Tushuntirish**: Agar 1-to'plamdagi barcha elementlar ikkinchi to'plamda mavjud bo'lsa ```True```(Rost), aks holda ```False```(yolg'on) qiymat qaytaradi.

**Umumiy sintaksis**:
```
toplam1.issubset(toplam2)
```

**Maxsus sintaksis**
```
toplam <= toplam2
```

Bu yerda:
- **toplam**: Majburiy parametr. 
- **tomplam2**: Majburiy parametr.

````python
# Oddiy hol:
x = {"a", "b", "c"}
y = {"f", "e", "d", "c", "b", "a"}

z = x.issubset(y)

print(z)

# -------------------------------------------------------------------------
# Tushuntirish:
#   - True qiymati qaytarilmoqda. Sababi x to'plamdagi barcha elementlar
#     y to'plamda mavjud.
````

````python
# Oddiy hol: Qisqa belgi bilan
x = {"a", "b", "c"}
y = {"f", "e", "d", "c", "b", "a"}

z = x <= y

print(z)
````

````python
# Maxsus hol: Agar barcha elementlar mavjud bo'lmasa.
x = {"a", "b", "c"}
y = {"f", "e", "d", "c", "b"}

z = x.issubset(y)

print(z)

# -------------------------------------------------------------------------
# Tushuntirish:
#   - False qiymati qaytarilmoqda. Sababi x to'plamdagi "a" belgisi y 
#     to'plamda mavjud emas.
````

<h2 id="set-symmetric_difference">14. symmetric_difference()</h2>

---

**Tushuntirish**: Ikki to'plam o'rtasidagi simmetrik ayirmani hisoblaydi. Ya'ni 1-to'plamning 2-to'plamda mavjud bo'lmagan va 2-to'plamning 1-to'plamda mavjud bo'lmagan elementlarini birlashtirib qaytaradi.

**Umumiy sintaksis**:
```
toplam.simmetric_difference(toplam2):
```

**Maxsus sintaksis**:
```
toplam ^ toplam2
```

Bu yerda:
-   **toplam**: Majburiy parametr.
-   **toplam2**: Majburiy parametr.

````python
# Oddiy hol:
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x ^ y

print(z) 
````

<h2 id="set-symmetric_difference_update">15. symmetric_difference_update()</h2>

---

**Tushuntirish**: Ikki to'plam o'rtasidagi simmetrik ayirmani hisoblaydi. Ya'ni 1-to'plamning 2-to'plamda mavjud bo'lmagan va 2-to'plamning 1-to'plamda mavjud bo'lmagan elementlarini birlashtirib natijasini birinchi to'plamga yozadi.

**Umumiy sintaksis**:
```
toplam.simmetric_difference_update(toplam2):
```

**Maxsus sintaksis**:
```
toplam ^= toplam2
```

Bu yerda:
-   **toplam**: Majburiy parametr.
-   **toplam2**: Majburiy parametr.

````python
# Oddiy hol:
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

x ^= y

print(x) 
````

<h2 id="set-union">16. union()</h2>

---

**Tushuntirish**: Ikki yoki undan ortiq to'plamlarni birlashtiradi va yangi to'plamni qaytaradi. Yangi to'plamdagi elementlar takrorlanmaydi.

**Umumoy sintaksis**:
```
toplam.union(toplam2, toplam3, ...)
```

**Maxsus sintaksis**:
```
toplam | toplam2 | toplam3 | ... 
```

Bu yerda:
-   **toplam**, **toplam2**: Majburiy elementlar.
-   **toplam3**, ... : Ixtiyoriy parametrlar.

````python
# Oddiy hol:
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x | y

print(z) 
````

````python
# Maxsus hol: 3 ta to'plamlarni birlashtirish
x = {"a", "b", "c"}
y = {"c", "d", "e"}
z = {"f", "g", "c"}

result = x | y | z

print(result)

# -------------------------------------------------------------------------
# Tushuntirish:
#   - Bu yerda ko'rinib turibdiki 3 ta to'plamda ham 'c' belgisi bor.
#     Ammo bu bir marta takrorlanmoqda.
````

<h2 id="set-update">17. update()</h2>

---

**Tushuntirish**: Ikki yoki undan ortiq to'plamlarni birlashtiradi va natijasini birinchi to'plamga yozadi. Yangi to'plamdagi elementlar takrorlanmaydi.

**Umumoy sintaksis**:
```
toplam.union(toplam2, toplam3, ...)
```

**Maxsus sintaksis**:
```
toplam |= toplam2 | toplam3 | ... 
```

Bu yerda:
-   **toplam**, **toplam2**: Majburiy elementlar.
-   **toplam3**, ... : Ixtiyoriy parametrlar.

````python
# Oddiy hol:
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

x |= y

print(x) 
````

````python
# Maxsus hol: 3 ta to'plamlarni birlashtirish
x = {"a", "b", "c"}
y = {"c", "d", "e"}
z = {"f", "g", "c"}

x |= y | z

print(x)

# -------------------------------------------------------------------------
# Tushuntirish:
#   - Bu yerda ko'rinib turibdiki 3 ta to'plamda ham 'c' belgisi bor.
#     Ammo bu bir marta takrorlanmoqda.
````
