<h2 id="list-append">1. append()</h2>

---

**Tushuntirish**: Ro'yxat(list)ning oxiriga yangi element qo'shadi. Hech qanday qiymat qaytarmaydi.

**Sintaksis**:
```
royxat.append(element)
```

Bu yerda:
-   **element**: Majburiy parametr. Pythondagi barcha standart ma'lumot turini qabul qilishi mumkin.

````python
# Oddiy hol: Ro'yxatga oddiy element qo'shish
royxat = ["Python", "C++", "HTML"]
element = "Java"
royxat.append(element)
royxat
````

````python
# Maxsus hol: Ro'yxat ichida ro'yxat qo'shish.
royxat = ["Python", "C++", "HTML"]
element = ["Java", "JavaScript"]
royxat.append(element)
royxat
````

---

<h2 id="list-count">4. count()</h2>

---

**Tushuntirish**: Ro'yxat ichidagi biror elementning sonini qaytaradi.

**Sintaksis**:
```
royxat.count(element)
```

Bu yerda:
-   **element**: Majburiy parametr. Ixtiyoriy turdagi o'zgaruvchini qabul qiladi.


````python
# Oddiy hol: Mavjud element sonini topish
royxat = [1, 2, 1, 4, 5, 1, 1]
element = 1
el_soni = royxat.count(element)
print(f"Ro'yxat ichida {element} elementi {el_soni} marta uchraydi")
````

````python
# Maxsus hol: Mavjud  bo'lmagan element sonini topish
royxat = [1, 2, 1, 4, 5, 1, 1]
element = 6
el_soni = royxat.count(element)
print(f"Ro'yxat ichida {element} elementi {el_soni} marta uchraydi")
````

<h2 id="list-extend">5. extend()</h2>

---

**Tushuntirish**: Joriy ro'yxatga yangi ro'yxat(qator, lug'at, to'plam va h.k)ni birlashtiradi. Hech qanday element qaytarmaydi.

**Sintaksis**:
```
royxat.extend(iterable)
```

Bu yerda:
-   **iterable**: Majburiy parametr. Ro'yxat, lug'at, qator, to'plam kabi ma'lumotlarni qabul qiladi.

````python
# Oddiy hol: Ikki ro'yxatni birlashtirish
royxat = ["a", "b", 3]
yangi_royxat = ["c", "d"]
royxat.extend(yangi_royxat)
print(royxat)
````

````python
# Maxsus hol: Ro'yxat bilan to'plam elementlarini birlashtirish
royxat = ["a", "b", 3]
yangi_toplam = {"c", "d"}
royxat.extend(yangi_toplam)
print(royxat)
````

````python
# Maxsus hol: Ro'yxat bilan qator elementlarini birlashtirish
royxat = ["a", "b", 3]
yangi_qator = ("c", "d")
royxat.extend(yangi_qator)
print(royxat)
````

````python
# Maxsus hol: Ro'yxat bilan lug'at elementlarini birlashtirish
royxat = ["a", "b", 3]
yangi_lugat = {"d": 1, "e": 2}
royxat.extend(yangi_lugat)
print(royxat)

# Tushuntirish:
#   - Ro'yxat bilan lug'at birlashtirilganda lug'atning kalitlari
#     olinadi.
````

<h2 id="list-index">6. index()</h2>

---

**Tushuntirish**: Berilgan ro'yxatda uchragan _qiymat_ ning birinchi indeksini qaytaradi. Agar _qiymat_ mavjud bo'lmasa xatolik kelib chiqadi.

**Sintaksis**:
```
royxat.index(value, start, stop)
```

Bu yerda:
- **value**(qiymat): Majburiy parametr. Faqat bilgilardan iborat ro'yxatni qabul qiladi. 
    Qidirilishi kerak bo'lgan ro'yxat.
- **start**(boshlanish): Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi.
    _qiymat_ ni qaysi indeksdan boshlab qidirishini belgilaydi. Jimlik bo'yicha 0.
- **stop**(tugash): Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi. 
    _qiymat_ ni qaysi indeksgacha qidirishini belgilaydi. 
    Jimlik bo'yicha ro'yxatning oxirgi indeksi

````python
# Oddiy hol: 
sonlar = [1, 11, 21, 11, 34]
qiymat = 11
index = sonlar.index(qiymat)
print(f"{qiymat} elementing ro'yxatdagi birinchi indeksi: {index}")
````

````python
# Maxsus hol: Qiymatni berligan oraliqda izlash(faqat boshlanish bilan)
sonlar = [1, 11, 21, 11, 34]
qiymat = 11
boshlanish = 1
index = sonlar.index(qiymat, boshlanish)
if index:

    print(f"{qiymat} elementing ro'yxatdagi birinchi indeksi: {index}")
````

````python
# Maxsus hol: Qiymatni berligan oraliqda izlash
sonlar = [1, 11, 21, 11, 34]
qiymat = 11
boshlanish = 1
tugash = 4
index = sonlar.index(qiymat, boshlanish, tugash)
if index:

    print(f"{qiymat} elementing ro'yxatdagi birinchi indeksi: {index}")
````

<h2 id="list-insert">7. insert()</h2>

---
<p style="text-align: justify;">
    <b>Tushuntirish</b>: Ro'yxatning ixtoyoriy joyiga yangi elementni joylashtiradi. Qolgan elementlar o'ng tomonga suriladi.
</p>

**Sintaksis**: 
```
royxat.insert(pos, elmt)
```

Bu yerda:
- **pos**: Majburiy parametr. Yangi elementni qayerga(qaysi indeksga) joylashtirishni bildiradi.
- **elmt**: Majburiy parametr. Qo'shilayotgan yangi element. Ixtiyoriy ma'lumot turini qabul qiladi.


````python
# Oddiy hol: 0 - indeks element joylashtirish
royxat = ["C++", "Java", "HTML", "PHP", "JS"]
element = "Python"
joylashuv = 0
royxat.insert(joylashuv, element)
print(royxat)

# Tushuntirish:
#   - Daslab ro'yxatda 5 ta element bor edi.
#   - 0 - indexda "C++" so'zi joylashgan edi.
#   - Keyin metod yordamida 0-indeksga "Python" 
#     so'zi qo'shildi va qolgan elementlar o'ng
#     tomonga surildi.
````

````python
# Oddiy hol: Ro'yxatning ixtiyoriy qismiga element qo'shish
royxat = ["C++", "Java", "HTML", "PHP", "JS"]
element = "Python"
joylashuv = 2
royxat.insert(joylashuv, element)
print(royxat)

# Tushuntirish:
#   - Daslab ro'yxatda 5 ta element bor edi.
#   - 2 - indexda "HTML" so'zi joylashgan edi.
#   - Keyin metod yordamida 2-indeksga "Python" 
#     so'zi qo'shildi va qolgan elementlar o'ng
#     tomonga surildi.
````

````python
# Maxsus hol: Ro'yxatning oxirgi qismiga element qo'shish
royxat = ["C++", "Java", "HTML", "PHP", "JS"]
element = "Python"
joylashuv = -1 # -1 ro'yxatning oxirini bildiradi.
royxat.insert(joylashuv, element)
print(royxat)

# Tushuntirish:
#   - Daslab ro'yxatda 5 ta element bor edi.
#   - Oxirgi indexda "JS" so'zi joylashgan edi.
#   - Keyin metod yordamida oxirgi indeksga 
#     "Python" so'zi qo'shildi va qolgan elementlar o'ng
#     tomonga surildi.
````

<h2 id="list-pop">8. pop()</h2>

---
**Tushuntirish**: Berilgan indeksdagi elementni ro'yxat tarkibidan o'chiradi va qiymat sifatida qaytaradi.

**Sintaksis**:
```
royxat.pop(pos)
```

Bu yerda:
- **pos**: Ixtiyoriy parametr. O'chirilayotgan elementning indeks raqamini bildiradi. Jimlik bo'yicha ```-1```.

````python
# Oddiy hol: Jimlik bo'yicha ishlash
royxat = ["C++", "Java", "HTML", "PHP", "JS"]
print(f"{royxat.pop()} so'zi ro'yxatdan olib tashlandi.\n\
Yangi ro'yxat: {royxat}")
````

````python
# Oddiy hol: Ixtiyoriy element o'chirib yangi o'zgaruvchida saqlash
royxat = ["C++", "Java", "HTML", "PHP", "JS"]
joylashuv = 2

x = royxat.pop(joylashuv)
print(f"{x} so'zi ro'yxatdan o'chirildi. Yangi ro'yxat:\n\
{royxat}")

# Tushuntirish:
#   - Bu yerda 2-indeksdagi 'HTML' so'zi o'chirildi va 
#     x o'zgaruvchisiga ta'minlandi.
````

<h2 id="list-remove">9. remove()</h2>

---
**Tushuntirish**: Berilgan qiymatga asosan ro'yxatdagi birinchi uchralgan elementni o'chiradi. Hech qanday qiymat qaytarmaydi.

**Sintaksis**:
```
royxat.remove(elmnt)
```

Bu yerda:
- **elmnt**: Majburiy parametr. Ixtiyoriy ma'lumot turini qabul qiladi. O'chirilayotgan elementni bildiradi.

````python
# Oddiy hol:
royxat = ["C++", "Python", "Java", "HTML", "PHP", "Python", "JS"]
element = "Python"
royxat.remove(element)

print(f"{element} so'zi ro'yxatdan olib tashlandi.\n\
Yangi ro'yxat: {royxat}")

# Tushuntirish:
#   - Ro'yxatda 'Python' so'zi 2 marta: 1-va 5- indekslarda
#     uchraydi.
#     'Python' so'zi birinchi bo'lib 1-indeksda mavjudligi
#     sababli o'chirildi.
````

<h2 id="list-reverse">10. reverse()</h2>

---
**Tushuntirish**: Ro'yxat elementlarini indekslariga teskari tartibda joylashtiradi. Hech qanday qiymat qaytarmaydi. Ya'ni:
- 0-indeksdagi element oxirgi indeksdagi element bilan almashiriladi.
- 1-indeksdagi element oxirgi indeksdan bitta oldigi element bilan almashtiriladi.
- va h.k.

**Sintaksis**:
```
royxat.reverse()
```

````python
# Maxsus hol: Sonlarni teskari tartibda joylashtirish
sonlar = [4, 5, 1, 3, 6]
sonlar.reverse()
print(f"Teskari saralangan ro'yxat: \n\
{sonlar}")
````

<h2 id="list-sort">11. sort()</h2>

---
**Tushuntirish**: Ro'yxat elementlarini qiymatlariga asosan saralaydi. Hech qanday qiymat qaytarmaydi.

**Sintaksis**: 
```
royxat.sort(reverse=True|False, key=myFunc)
```

Bu yerda:
- **reverse**: Ixtiyoriy parametr. Ro'yxat elementlarini teskari tartibda saralashni bildiradi, agar ```True``` qiymatni qabul qilsa. Jimlik bo'yicha ```False```.
- **key**: Ixtiyoriy parametr. Python funksiyasini qabul qiladi. Ro'yxatni qaysi qoidaga asoslanib saralashni bildiradi.

````python
# Maxsus hol: Sonlarni kamayish tartibda joylashtirish
sonlar = [4, 5, 1, 3, 6]
sonlar.sort(reverse=True)
print(f"Kamayish tartida saralangan ro'yxat: \n\
{sonlar}")
````

````python
# Maxsus hol: So'zlarni o'sih tartibda joylashtirish
sozlar = ["Salom", "dunyo", "Python"]
sozlar.sort() 
print(f"O'sish tartida saralangan ro'yxat: \n\
{sozlar}")
````

````python
# Maxsus hol: So'zlarning uzunliga qarab o'sish tartibida saralash

# Bu funksiya so'zning uzunligini qaytaradi.
def myFunc(e):
  return len(e)

mashinalar = ['Ford', 'Mitsubishi', 'BMW', 'VW']
mashinalar.sort(key=myFunc)
mashinalar

# Tushuntirish:
#   - 'myFunc' funksiyasi mos ravishda har bir
#     so'zning uzunligini hisoblayapti va ular
#     quyidagicha: [4, 10, 3, 2]
#   - Bu ro'yxatni o'sish tartibida saralaganda
#     quyidagicha bo'ladi: [2, 3, 4, 10]
#   - Natija esa har bir so'zning uzunligiga 
#     asoslanib chiqmoqda.
````

````python
# Maxsus hol: Lug'atlardan iborat ro'yxatni barcha lug'atlarning
# 'year' kalit so'zini qiymati orqali saralash.

# Bu funksiya lug'atning 'year' kalitidagi qiymatini qaytaradi.
def myFunc(e):
  return e['year']

mashinalar = [
  {'car': 'Ford', 'year': 2005},
  {'car': 'Mitsubishi', 'year': 2000},
  {'car': 'BMW', 'year': 2019},
  {'car': 'VW', 'year': 2011}
]

mashinalar.sort(key=myFunc) 
mashinalar

# Tushuntirish:
#   - 'myFunc' funksiyasi mos ravishda har bir
#     lug'atning 'year' kalit so'zidagi qiymatini
#     qaytaradi va ular quyidagicha: 
#     [2005, 2000, 2019, 2011]
#   - Bu ro'yxatni o'sish tartibida saralaganda
#     quyidagicha bo'ladi: [2000, 2005, 2011, 2019]
#   - Natija esa har bir lug'atning 'year' kalitidagi
#     qiymatiga asoslanib chiqmoqda.
````
