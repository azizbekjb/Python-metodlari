<h2 id="append">1. append()</h2>

---

**Tushuntirish**: Ro'yxat(list)ning oxiriga yangi element qo'shadi.

**Sintaksis**:
```
royxat.append(element)
```

Bu yerda:
-   **element**: Majburiy parametr. Pythondagi barcha standart ma'lumot turini qabul qilishi mumkin.

````python
# Oddiy hol: Ro'yxatga oddiy element qo'shish
royxat = ["Python", "C++", "HTML"]
element = "Java"
royxat.append(element)
royxat
````

````python
# Maxsus hol: Ro'yxat ichida ro'yxat qo'shish.
royxat = ["Python", "C++", "HTML"]
element = ["Java", "JavaScript"]
royxat.append(element)
royxat
````
