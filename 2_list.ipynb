<h2 id="list-append">1. append()</h2>

---

**Tushuntirish**: Ro'yxat(list)ning oxiriga yangi element qo'shadi. Hech qanday qiymat qaytarmaydi.

**Sintaksis**:
```
royxat.append(element)
```

Bu yerda:
-   **element**: Majburiy parametr. Pythondagi barcha standart ma'lumot turini qabul qilishi mumkin.

````python
# Oddiy hol: Ro'yxatga oddiy element qo'shish
royxat = ["Python", "C++", "HTML"]
element = "Java"
royxat.append(element)
royxat
````

````python
# Maxsus hol: Ro'yxat ichida ro'yxat qo'shish.
royxat = ["Python", "C++", "HTML"]
element = ["Java", "JavaScript"]
royxat.append(element)
royxat
````

<h2 id="list-clear">2. clear()</h2>

---

**Tushuntirish**: Ro'yxatning barcha elementlarini o'chiradi. Hech qanday qiymat qaytarmaydi.

**Sintaksis**:
```
royxat.clear()
```


````python
# Misol
royxat = ["Python", "C++", "HTML"]
royxat.clear()
print(royxat)
````

<h2 id="list-copy">3. copy()</h2>

---

**Tushuntirish**: Ro'yxatning nusxasini qaytaradi.

**Sintaksis**: 
```
royxat.copy()
```


````python
royxat = ["Python", "C++", "HTML"]
nusxa = royxat.copy()
print(f"Asl ro'yxat: {royxat}")
print(f"Nusxa ro'yxat: {nusxa}")
````

<h2 id="list-count">4. count()</h2>

---

**Tushuntirish**: Ro'yxat ichidagi biror elementning sonini qaytaradi.

**Sintaksis**:
```
royxat.count(element)
```

Bu yerda:
-   **element**: Majburiy parametr. Ixtiyoriy turdagi o'zgaruvchini qabul qiladi.


````python
# Oddiy hol: Mavjud element sonini topish
royxat = [1, 2, 1, 4, 5, 1, 1]
element = 1
el_soni = royxat.count(element)
print(f"Ro'yxat ichida {element} elementi {el_soni} marta uchraydi")
````

````python
# Maxsus hol: Mavjud  bo'lmagan element sonini topish
royxat = [1, 2, 1, 4, 5, 1, 1]
element = 6
el_soni = royxat.count(element)
print(f"Ro'yxat ichida {element} elementi {el_soni} marta uchraydi")
````

<h2 id="list-extend">5. extend()</h2>

---

**Tushuntirish**: Joriy ro'yxatga yangi ro'yxat(qator, lug'at, to'plam va h.k)ni birlashtiradi. Hech qanday element qaytarmaydi.

**Sintaksis**:
```
royxat.extend(iterable)
```

Bu yerda:
-   **iterable**: Majburiy parametr. Ro'yxat, lug'at, qator, to'plam kabi ma'lumotlarni qabul qiladi.

````python
# Oddiy hol: Ikki ro'yxatni birlashtirish
royxat = ["a", "b", 3]
yangi_royxat = ["c", "d"]
royxat.extend(yangi_royxat)
print(royxat)
````

````python
# Maxsus hol: Ro'yxat bilan to'plam elementlarini birlashtirish
royxat = ["a", "b", 3]
yangi_toplam = {"c", "d"}
royxat.extend(yangi_toplam)
print(royxat)
````

````python
# Maxsus hol: Ro'yxat bilan qator elementlarini birlashtirish
royxat = ["a", "b", 3]
yangi_qator = ("c", "d")
royxat.extend(yangi_qator)
print(royxat)
````

````python
# Maxsus hol: Ro'yxat bilan lug'at elementlarini birlashtirish
royxat = ["a", "b", 3]
yangi_lugat = {"d": 1, "e": 2}
royxat.extend(yangi_lugat)
print(royxat)

# Tushuntirish:
#   - Ro'yxat bilan lug'at birlashtirilganda lug'atning kalitlari
#     olinadi.
````

<h2 id="list-index">6. index()</h2>

---

**Tushuntirish**: Berilgan ro'yxatda uchragan _qiymat_ ning birinchi indeksini qaytaradi. Agar _qiymat_ mavjud bo'lmasa xatolik kelib chiqadi.

**Sintaksis**:
```
royxat.index(value, start, stop)
```

Bu yerda:
- **value**(qiymat): Majburiy parametr. Faqat bilgilardan iborat ro'yxatni qabul qiladi. 
    Qidirilishi kerak bo'lgan ro'yxat.
- **start**(boshlanish): Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi.
    _qiymat_ ni qaysi indeksdan boshlab qidirishini belgilaydi. Jimlik bo'yicha 0.
- **stop**(tugash): Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi. 
    _qiymat_ ni qaysi indeksgacha qidirishini belgilaydi. 
    Jimlik bo'yicha ro'yxatning oxirgi indeksi

````python
# Oddiy hol: 
sonlar = [1, 11, 21, 11, 34]
qiymat = 11
index = sonlar.index(qiymat)
print(f"{qiymat} elementing ro'yxatdagi birinchi indeksi: {index}")
````

````python
# Maxsus hol: Qiymatni berligan oraliqda izlash(faqat boshlanish bilan)
sonlar = [1, 11, 21, 11, 34]
qiymat = 11
boshlanish = 1
index = sonlar.index(qiymat, boshlanish)
if index:

    print(f"{qiymat} elementing ro'yxatdagi birinchi indeksi: {index}")
````

````python
# Maxsus hol: Qiymatni berligan oraliqda izlash
sonlar = [1, 11, 21, 11, 34]
qiymat = 11
boshlanish = 1
tugash = 4
index = sonlar.index(qiymat, boshlanish, tugash)
if index:

    print(f"{qiymat} elementing ro'yxatdagi birinchi indeksi: {index}")
````
