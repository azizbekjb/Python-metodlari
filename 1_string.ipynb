## 1. capitalize()
---

**Tushuntirish**: Matnning birinchi belgisini katta harfga o'zgartiradi, agar raqam bo'lsa o'zgarishsiz qoladi.

**Sintaksis**:
```
"mening matnim".capitalize()
```

````python
# Oddiy matn
matn = "men Azizbek Jabborov Python dasturchisiman."
yangi_matn = matn.capitalize()
print("Eski matn:", matn)
print("Yangi matn:", yangi_matn)
````

````python
# Birinchi belgi raqam bo'lsa
raqamli_matn = "18 soni 9 ga bo'linadi."
yangi_raqamli_matn = raqamli_matn.capitalize()
print("Eski matn:", raqamli_matn)
print("Yangi matn:", yangi_raqamli_matn)
````

## 3. center()
---

**Tushuntirish**: Berilgan matnni eksi o'lchovidan _uzunlik_ o'lchoviga o'zgartirib, matnni markazlashtiradi. Matnning chap va o'ng tomonini kiritilgan _qiymat_ bilan to'ldiriladi. Agar qiymat berilmasa bo'sh joy(" ") bilan to'ldiradi.

**Sintaksis**:
```
matn.center(uzunlik, qiymat)
```

Bu yerda:
- **uzunlik**: Majburiy parametr. Agar bu parametr kirtilmasa xatolik kelib chiqadi.
- **qiymat**: Ixtiyoriy parametr. Matnning chap va o'ng tomoniga _uzunlik_ marta to'ldiriladi.


````python
# Oddiy hol. Bo'sh jo'y bilan to'ldirish
matn = "mevalar"
yangi_matn = matn.center(10) 
print("Eski matn:", matn, "Uzunligi:", len(matn)) # Eski matnnig uzunligi 7.
print("Yangi matn:", yangi_matn, "Uzunligi:", len(yangi_matn)) # Matn markazlashtirildi va 7 o'lchovli matndan bo'sh qiymat(" ") yordamida 10 o'lchovli matnga o'zgaritirildi.
````

````python
yangi_matn
````

````python
# Maxsus bilgi bilan to'ldirish
matn = "mevalar"
yangi_matn = matn.center(10, "A") 
print("Eski matn:", matn, "Uzunligi:", len(matn)) # Eski matnnig uzunligi 7.
print("Yangi matn:", yangi_matn, "Uzunligi:", len(yangi_matn)) # Matn markazlashtirildi va 7 o'lchovli matndan "A" belgisi yordamida 10 o'lchovli matnga o'zgaritirildi.
````

## 4. count()
---

**Tushuntirish**: Berilgan matndan barchan uchraga _qiymat_ sonini qaytaradi.

**Sintaksis**:
```
matn.count(qiymat, boshlanish, tugash)
```

Bu yerda:
- **qiymat**: Majburiy parametr. Faqat bilgilardan iborat matnni qabul qiladi. Qidirilishi kerak bo'lgan matn.
- **boshlanish**: Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi. _qiymat_ ni qaysi indeksdan boshlab qidirishini belgilaydi. Jimlik bo'yicha 0.
- **tugash**: Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi. _qiymat_ ni qaysi indeksgacha qidirishini belgilaydi. Jimlik bo'yicha matnning oxirgi indeksi.

````python
# Oddiy hol.
matn = "Men olmalarni juda yaxshi ko'raman va olmalar juda mazali meva."
olma_soni = matn.count("olma")
print("Berilgan matndagi 'olma' soni:", olma_soni, "ta.")
````

````python
# Maxsus hol. 'olma' so'zini 10- va 50- indekslar orasidan izlash.
matn = "Men olmalarni juda yaxshi ko'raman va olmalar juda mazali meva."
qiymat = "olma"
boshlanish = 10
tugash = 50
olma_soni = matn.count(qiymat, boshlanish, tugash)
print(f"{boshlanish}-indeksdan {tugash}-indeksigacha bo'lgan qismdan 'olma' soni:", olma_soni, "ta.")
````

## 5. encode()
---

**Tushuntirish**: Berilgan matnni kodga o'giradi.

**Sintaksis**:
```
matn.encode(encoding, errors)
```

Bu yerda:
- **encoding**: Ixtiyoriy parametr. Matnli(string) qiymat qabul qiladi. Jimlik bo'yicha 'UTF-8'.
- **errors**: Ixtiyoriy parametr. Matnli(string) qiymat qabul qiladi. 

    Agar matndagi ma'lum bir belgi kodlashtirilmasa, yuzaga keladigan xatolikni oldinini olsih uchun ishlatiladi. 

    Bu parametr quyidagi qiymatlarni qabul qiladi:
    - ```'backslashreplace'```: Kodlashtirilmagan belgi o'rniga bekslesh(\\) belgisini qo'yadi.
    - ```'ignore'```: Kodlashtirilmagan belgi matnnda olib tashlaniladi.
    - ```'namereplace'```: Kodlashtirilmagan belgi o'rniga o'sha belgining tavsifini joylaydi.
    - ```'strict'```: Jimlik bo'yicha qiymat. Xatolik keltirib chiqaradi.
    - ```'replace'```: Kodlashtirilmagan belgi o'rniga so'roq belgisi(?) belgisini qo'yadi.
    - ```'xmlcharrefreplace'```: Kodlashtirilmagan belgi o'rniga xml belgisini qo'yadi.

````python
txt = "My name is Ståle"

print(txt.encode(encoding="ascii",errors="backslashreplace"))
print(txt.encode(encoding="ascii",errors="ignore"))
print(txt.encode(encoding="ascii",errors="namereplace"))
print(txt.encode(encoding="ascii",errors="replace"))
print(txt.encode(encoding="ascii",errors="xmlcharrefreplace"))
````

## 6. endswith()
---

**Tushuntirish**: Agar matn berilgan _qiymat_ bilan tugasa rost(True) qaytaradi, aks holda yolg'on(False).

**Sintaksis**:
```
matn.endswith(qiymat, boshlanish, tugash)
```

Bu yerda:
- **qiymat**: Majburiy parametr. String(matn) yoki tuple(qator) bo'lishi mumkin. Agar tuple kiritilgan bo'lsa va matnning oxiri tuple ichidagi birorta element bilan tugasa rost qaytaradi.
- **boshlanish**: Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi. _qiymat_ ni qaysi indeksdan boshlab qidirishini belgilaydi. Jimlik bo'yicha 0.
- **tugash**: Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi. _qiymat_ ni qaysi indeksgacha qidirishini belgilaydi. Jimlik bo'yicha matnning oxirgi indeksi.



````python
# Oddiy hol
matn = "Mening ismim Azizbek"
qism = "bek"
natija = matn.endswith(qism)

if natija:
    print(f"Berilgan matn '{qism}' so'zi bilan tugaydi.")
else:
    print(f"Berilgan matn '{qism}' so'zi bilan tugamaydi.")
````

````python
# Maxsus hol: Kerakli qismdan izlash
matn = "Mening ismim Azizbek"
qism = "bek"
boshlanish = 10
tugash = len(matn) - 2
natija = matn.endswith(qism, boshlanish, tugash)

if natija:
    print(f"Berilgan matn {boshlanish}-indeksdan  {tugash - 1}-indeksi oralig'ida '{qism}' so'zi bilan tugaydi.")
    print(f"Kerakli qism: {matn[boshlanish: tugash]}")
else:
    print(f"Berilgan matn {boshlanish}-indeksdan  {tugash - 1}-indeksi oralig'ida '{qism}' so'zi bilan tugamaydi.")
    print(f"Kerakli qism: {matn[boshlanish: tugash]}")
````

````python
# Maxsus hol.
matn = "Mening ismim Azizbek"
qismlar = ("bek", "zbek")
natija = matn.endswith(qismlar)

if natija:
    print(f"Berilgan matn '{qismlar}' ichidagi biror so'z bilan tugaydi.")
else:
    print(f"Berilgan matn '{qismlar}' ichidagi biror so'z bilan tugamaydi.")
````

## 7. expnadtabs()
---

**Tushuntirish**: Matndagi tab(\t)lar o'rniga ma'lum bir son miqdor bo'sh joy(" ") bilan o'zgartiradi.

**Sintaksis**:
```
matn.expandtabs(tabsize)
```

Bu yerda:
-   **tabsize**: Ixtiyoriy parametr. Butun son(integer) qabul qiladi. Tab belgisi o'rniga nechta bo'sh qo'yishni bildiradi. Jimlik bo'yicha 7.

    Misol uchun agar ```tabsize=7``` bo'lsa, matndagi har bitta tab belgisi o'rniga ```6``` bo'sh joy qo'yiladi.

**Eslatma**: ```Tab``` - bu 4 bo'sh joydan iborat matn. Pythonda tabning uzunligi 1 ga teng. Quyidagi kod orqali tekshirib ko'ring.
```
len("\t") # Natijasi 1 chiqadi. 
```


````python
matn = "M\te\ti\tn\ti\tn\tg"
print(f"Haqiqiy matn. Uzunligi: {len(matn)} ga teng.", ) # Haqiqiy matn
print(matn)

# Tushutirish:
#   - Matnda jami bo'lib 7 harf va 6 ta tab bor.
#   - Shunday ekan matnning uzunligi 13 ga teng bo'ladi.
````

````python
print(matn.expandtabs(3)) # Matndagi barcha tab belgisi o'rniga 2 ta bo'sh joy qo'yish
print(matn.expandtabs()) #  Matndagi barcha tab belgisi o'rniga 7 ta(jimlik bo'yicha) bo'sh joy qo'yish.
print(matn.expandtabs(10)) # Matndagi barcha tab belgisi o'rniga 9 ta bo'sh joy qo'yish

# Tushuntirish:
#   - Oxirgi misolda tabsize=10 kiritdik.
#   - Matnnig 0-indeksida 'M' belgisi, 10-indeksida esa 'e' belgisi joylashgan.
#   - 'M' va 'e' belgilari o'rtasida 9 ta bo'sh joy(probel) bor.
````

## 8. find()
---

**Tushuntirish**: Matnda maxsus _qiymat_ borligini tekshradi. Agar mavjud bo'lsa uchralgan birinchi indeksni qaytardi, aks holda ```-1```.

**Sintaksis**:
```python
matn.find(value, start, stop)
```

Bu yerda:
- **value**(qiymat): Majburiy parametr. Faqat bilgilardan iborat matnni qabul qiladi. Qidirilishi kerak bo'lgan matn.
- **start**(boshlanish): Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi. _qiymat_ ni qaysi indeksdan boshlab qidirishini belgilaydi. Jimlik bo'yicha 0.
- **stop**(tugash): Ixtiyoriy parametr. Butun son(integer)ni qabul qiladi. _qiymat_ ni qaysi indeksgacha qidirishini belgilaydi. Jimlik bo'yicha matnning oxirgi indeksi.


````python
# Oddiy hol
matn = "Mening ismim Azizbek"
qism = "A"
natija = matn.find(qism)

if natija != -1:
    print(f"'{qism}' so'zi '{matn}' matnida mavjud. U {natija}-indeksida uchraydi.")
else:
    print(f"'{qism}' so'zi '{matn}' matnida mavjud emas.")
````

````python
# Maxsus hol
matn = "Mening ismim Azizbek"
qism = "i"
boshlanish = 2
tugash = 10
natija = matn.find(qism, boshlanish, tugash)

if natija != -1:
    print(f"'{qism}' so'zi '{matn[boshlanish: tugash]}' matnida mavjud. U {natija}-indeksida uchraydi.")
else:
    print(f"'{qism}' so'zi '{matn[boshlanish: tugash]}' matnida mavjud emas.")

# Tushuntirish:
#   - Matndan 'i' belgisi 2- va 10- indekslar orasida qidirilmoqda,
#   - 'i' belgisi matnda birinchi bo'lib 3 - indeksda(4-xonada) mavjud. 
````

## 9. format()
---

**Tushuntirish**: Matnnig ichiga maxsus usul bilan qiymatlar joylanadi.

**Sintaksis**:
```
matn.format(value1, value2, ...)
```

Bu yerda:
-   **value1**, **value2**, ... : Majburiy parametrlar. Kalit=qiymat usuli yoki vergul bilan ajratilgan holda indekslar yordamida to'ldirish mumkin.

````python
matn1 = "Mening ismim {ism}. Yoshim {yosh} da".format(ism="Azizbek", yosh=20) # kalit=qiymat usuli bilan to'ldirish
matn2 = "Mening ismim {0}. Yoshim {1} da".format("Azizbek", 20) # Vergul bilan ajratilgan holda indeks bilan to'ldirish.
matn3 = "Mening ismim {}. Yoshim {} da".format("Azizbek", 20) # Vergul bilan ajratilgan holda ketma-ket to'ldirish. Jimlik bo'yicha.

print(matn1)
print(matn2)
print(matn3)
````

````python
# Maxsus hollar.
matn4 = "Siz to'pladingiz {:%}" # vergul bilan xona aniqligida ko'rsatish.
print(matn4.format(0.215))

matn5 = "Siz to'pladingiz {:.0%}" # Hech qanday aniqliksiz.
print(matn5.format(0.25))
````

## 10. format_map()
---

**Tushuntirish**: Lug'at(dict) yordamida matnni formatlash usuli.

**Sintaksis**: 
```

```

Bu yerda:
-   **dictionary**(lug'at): Majburiy parametr. Faqat lug'at(dict) ma'lumot turini qabul qiladi. Matnning kerakli qismini lug'atdagi kalitlar va qiymatlar yordamida to'ldirish uchun ishlatiladi.
