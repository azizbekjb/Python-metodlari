<h2 id="dict-clear">1. clear()</h2>

---
**Tushuntirish**: Lug'atning barcha elementlarini o'chiradi. Hech qanday qiymat qaytarmaydi.

**Sintaksis**:
```
lugat.clear()
```

````python
# Oddiy hol: Lug'atning barcha elementlarini o'chirish.
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

mashina.clear()
print(mashina)
````

lugat.copy()

<h2 id="dict-fromkeys">3. fromkeys()</h2>

---
**Tushuntirish**: Maxsus kalit va qiymatlardan yangi lug'at hosil qiladi.

**Sintaksis**:
```
dict.fromkeys(keys, value)
```

Bu yerda: 
- **keys**(kalitlar): Majburiy parametr. Ixtiyoriy iteratsiyalovchi(ro'yxat, qator, to'plam va ...) qiymatlarni qabul qiladi.
- **value**(qiymat): Ixtiyori parametr. Har bir kalitga mos yagona qiymat. Jimlik bo'yicha ```None``` 

````python
# Oddiy hol: Jimlik qoidasi bo'yicha ishlash.
x = ("kalit1", "kalit2", "kalit3")
lugat = dict.fromkeys(x)
print(lugat)

# Tushuntirish:
#   - Jimlik qoidasi bo'yicha har bir kalitga None
#     qiymati qo'yilmoqda.
````

````python
# Maxsus hol: Kerakli qiymatni berish.
x = ("kalit1", "kalit2", "kalit3")
y = 0
lugat = dict.fromkeys(x, y)
print(lugat)

# Tushuntirish:
#   - Har bir kalitga 0 qiymati qo'yilmoqda.
````

<h2 id="dict-get">4. get()</h2>

---
**Tushuntirish**: Kerakli kalitga mos qiymatni qaytaradi.

**Sintaksis**:
```
lugat.get(keyname, value)
```

Bu yerda:
- **keyname**(kalit): Majburiy parametr. Lug'atdagi kerakli kalit nomini bildiradi.
- **value**(qiymat): Ixtiyoriy parametr. Agar kalit lug'atda mavjud bo'lmasa, metod shu qiymatni qaytaradi. Jimlik bo'yicha ```None```

````python
# Oddiy hol: 'model' kalitiga mos qiymatni qaytarish.
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

print(mashina.get("model"))
````

````python
# Maxsus hol: Jimlik qoidasi bo'yicha ishlash.
# Agar kalit lug'atda mavjud bo'lmasa
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalit = "narxi"
print(lugat.get(kalit))
````

````python
# Maxsus hol: Maxsus qiymat bilan ishlash
# Agar kalit lug'atda mavjud bo'lmasa.

mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalit = "narxi"
narx = 15000
print(f"Mashina narxi: {lugat.get(kalit, narx)}$")
````

<h2 id="dict-items">5. items()</h2>

---
**Tushuntirish**: Lug'atning har bir kalit va qiymatini bitta qator ichiga joylaydi, bir nechta qatorlardan iborat ro'yxat qaytatadi.

**Sintaksis**:
```
lugat.items()
```

````python
# Oddiy hol: Kalit-qiymat juftligini qaytarish.
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

items = mashina.items()
print(items)

# Tushuntirish:
#   - Har bir kalit-qiymat alohida qator ichiga
#     solinmoqda va barcha qatorlar bitta ro'yxat
#     ichida qaytarilmoqda.
````

````python
# Maxsus hol
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

items = mashina.items()

mashina["yili"] = 2018
print(items)

# Tushutirish:
#   - Agar lug'atdagi biror bir kalitning qiymati
#     o'zgartirlsa, 'items()' metodining qiymati 
#     ham o'zgaradi.
````

<h2 id="dict-pop">7. pop()</h2>

---
**Tushuntirish**: Berilgan kalitni qiymati bilan lug'atdan o'chiradi va yangi qiymat sifatida o'sha kalitning qiymatini qaytaradi.

**Sintaktsis**:
```
lugat.pop(keyname, defaultvalue)
```

Bu yerda:
- **keyname**(kalit nomi): Majburiy parametr. O'chirish uchun kerakli kalit nomi.
- **defaultvalue**(standart qiymat): Ixtiyoriy parametr. Agar berilgan kalit nomi lug'atda 
mavjud bo'lmasa, shu qiymat qaytariladi.

    Agar bu parametrga qiymat berilmasa va berilgan kalit nomi lug'atda mavjud bo'lsa xatolik kelib chiqadi.

````python
# Oddiy hol: Mavjud kalitni o'chirish
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalit= "yili"
mashina.pop(kalit)
print(mashina)
````

````python
# Oddiy hol: Mavjud kalitni o'chirish va yangi
# o'zgaruvchiga taminlash
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalit= "yili"
x = mashina.pop(kalit)
print(f"{kalit} ning qiymati: {x}")

# Tushuntirish:
#   - Yuqoridagi qoidaga qarang.
````

````python
# Maxsus hol: Mavjud bo'lmaga kalitni o'chirish
# va boshqa qiymatni qaytarish
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalit = "narxi"
standart_qiymat = 15000
print(mashina.pop(kalit, standart_qiymat))

# Tushuntirish:
#   - 'narxi' nomli kalit so'z lug'atda
#      mavjud emas, shu sababli kiritilga
#      standart qiymat qaytarilmoqda.
````

````python
# Maxsus hol: Mavjud bo'lmaga kalitni o'chirish
# hech qanday standart qiymatsiz ishlatish
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalit = "narxi"
try:
    print(mashina.pop(kalit))
except KeyError:
    print(f"{kalit} nomli kalit so'z lug'atda mavjud emas.")
````

<h2 id="dict-setdefault">9. setdefault()</h2>

---
**Tushuntirish**: Berilgan kalitga mos qiymatni qaytaradi. Agar kalit mavjud bo'lmasa, berilgan _kalit_ lug'atga _maxsus qiymat_ bilan yangi element sifatida qo'shiladi.

**Sintaksis**:
```
lugat.setdefault(keyname, value)
```

Bu yerda:
- **keyname**(kalit nomi): Majburiy parametr. Kerakli qiymatni olish uchun kalit so'z.
- **value**(qiymat): Ixtiyoriy parametr. Jimlik bo'yicha ```None```.

    Agar kalit so'z lug'atda mavjud bo'lsa, bu parametr natjidga ta'sir qilmaydi

    Agar kalit so'z lug'atda mavjud bo'lmasa, yangi kalit-qiymat juftlig lug'atga qo'shiladi.

````python
# Oddiy hol: Mavjud kalitdan qiymatni olish
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalit = "yili"
standart_qiymat = 1950
print(f"Mashina {mashina.setdefault(kalit, standart_qiymat)}\
-yilda ishlab chiqarilgan")

# Tushuntirish:
#   - 'yili' nomli kalit so'z lug'atda mavjud.
#      Shu sababli uning qiymati qaytarilmoqda.
````

````python
# Oddiy hol: Mavjud bo'lmagan kalitdan qiymatni
# olish
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalit = "narxi"
standart_qiymat = 15000
print(
f"Mashinaning narxi: \
{mashina.setdefault(kalit, standart_qiymat)}$\n",
f"Mashina haqida umumiy ma'lumot: \n{mashina}")

# Tushuntirish:
#   - 'narxi' nomli kalit so'z lug'atda mavjud
#      emas. Shu sababli yangi 'narxi': 15000
#      elementi lug'atda qo'shilmoqda.
````

<h2 id="dict-update">10. update()</h2>

---
**Tushuntirish**: Lug'atning oxiriga yangi kalit-qiymat juftligni qo'shadi.

**Sintaksis**:
```
lugat.update(element)
```

Bu yerda:
- **element**: Majburiy parametr. Yangi kalit-qiymat juftligi.

````python
# Oddiy hol
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

yangi_element = {"narxi": 15000}
mashina.update(yangi_element)
print(mashina)
````

<h2 id="dict-values">11. values()</h2>

---
**Tushuntirish**: Lug'atning barchi kalitlarining qiymatlarini bitta ro'yxat ichida qaytaradi.

**Sintaksis**:
```
lugat.values()
```

````python
# Oddiy hol
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

print(mashina.values())
````

````python
# Maxsus hol: Yangi kalit qo'shish.
mashina = {
    "brend": "Ford",
    "model": "Mustang",
    "yili": 1964,
}

kalitlar = mashina.values()

mashina["narxi"] = 15000

print(kalitlar)

# Tushutirish:
#   - Agar lug'atga yangi kalit qo'shilsa 'values()' 
#     metodining qiymati ham o'zgaradi.
````
